Your program must:

Identify and count even and odd numbers.

Calculate the sum of all positive numbers.

Count negative numbers.

Identify the largest and smallest numbers.

Calculate the average of all entered numbers.

Skip zero values when calculating the sum of positive numbers using continue.

```python
even = 0
odd = 0

for number in range(11, 22):
    if number % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even Numbers:", even)
print("Odd Numbers:", odd)
numbers = [10, -5, 20, -3, 0, 15]
total = 0

for number in numbers:
    if number > 0:
        total += number

print("Sum of Positive Numbers:", total)
numbers = [10, -5, 20, -3, 0, -8]
count = 0

for number in numbers:
    if number < 0:
        count += 1

print("Negative Numbers:", count)
numbers = [10,12,-12,-45,23,-21]
count = 0
numbers = [10, 0, 20, 0, 30]
total = 0

for number in numbers:
    if number != 0:
        total += number

print("Sum:", total)
numbers = [10, -5, 20, -3, 0, 15]

print("Numbers:", numbers)
print("Total Numbers:", len(numbers))
print("Sum:", sum(numbers))
print("Largest:", max(numbers))
print("Smallest:", min(numbers))
print("Average:", sum(numbers) / len(numbers))
```
